```python
dept_attrition = df.groupby("Department")["Attrition"].mean() * 100
dept_attrition
```

```python
plt.figure(figsize=(7,5))

sns.barplot(x=dept_attrition.index, y=dept_attrition.values)

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")

plt.show()
```

```python
role_attrition = df.groupby("JobRole")["Attrition"].mean() * 100
role_attrition.sort_values(ascending=False)
```

```python
plt.figure(figsize=(10,6))

sns.barplot(x=role_attrition.values, y=role_attrition.index)

plt.title("Attrition Rate by Job Role")
plt.xlabel("Attrition Rate (%)")
plt.ylabel("Job Role")

plt.show()
```

```python
bins = [18,25,35,45,55,65]

labels = ["18-25","26-35","36-45","46-55","55+"]

df["AgeGroup"] = pd.cut(df["Age"], bins=bins, labels=labels)
```

```python
plt.figure(figsize=(7,5))

sns.barplot(x=age_attrition.index, y=age_attrition.values)

plt.title("Attrition Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Attrition Rate (%)")

plt.show()
```

```python
bins = [0,3,7,12,20,40]

labels = ["0-3","4-7","8-12","13-20","20+"]

df["TenureGroup"] = pd.cut(df["YearsAtCompany"], bins=bins, labels=labels)
```

```python
dept_risk = df.groupby("Department").agg(
    TotalEmployees=("Attrition","count"),
    EmployeesLeft=("Attrition","sum")
)

dept_risk["AttritionRate (%)"] = (dept_risk["EmployeesLeft"] / dept_risk["TotalEmployees"]) * 100

dept_risk.sort_values("AttritionRate (%)", ascending=False)
```

```python
role_risk = df.groupby("JobRole").agg(
    TotalEmployees=("Attrition","count"),
    EmployeesLeft=("Attrition","sum")
)

role_risk["AttritionRate (%)"] = (role_risk["EmployeesLeft"] / role_risk["TotalEmployees"]) * 100

role_risk.sort_values("AttritionRate (%)", ascending=False)
```
